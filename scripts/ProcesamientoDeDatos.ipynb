```python
import pandas as pd

# Carga del archivo CSV
data = pd.read_csv("../bd/bdlibros.csv")

# Inicializamos el contador
count_consecutive_values = 0

# Iteramos sobre las filas del DataFrame
for i in range(len(data) - 1):
    # Verificamos si hay un número en la columna 'Nº' en la fila actual y en la siguiente fila
    if pd.notna(data.loc[i, 'Nº']) and pd.notna(data.loc[i + 1, 'Nº']):
        count_consecutive_values += 1

# Mostrar el conteo de valores consecutivos
print(f"Cantidad de veces que hay un dato en la columna 'Nº' y en la siguiente fila también: {count_consecutive_values}")
```

```python
import pandas as pd

# Carga del archivo CSV
data = pd.read_csv("../bd/bdlibros.csv")

# Inicializamos la lista de resultados
extracted_records = []

# Iteramos sobre las filas del DataFrame
for i in range(len(data)):
    # Verificamos si hay un número en la columna 'Nº'
    if pd.notna(data.loc[i, 'Nº']):
        # Obtenemos el número de la columna 'Nº'
        numero = data.loc[i, 'Nº']
        
        # Obtenemos el contenido de 'ASIENTO BIBLIOGRAFICO'
        asiento_bibliografico = data.loc[i, 'ASIENTO BIBLIOGRAFICO']
        
        # Convertimos a cadena de texto y extraemos el autor de la fila actual
        author = str(asiento_bibliografico)
        
        # Convertimos a cadena de texto y extraemos el título de la siguiente fila
        title = str(data.loc[i + 1, 'ASIENTO BIBLIOGRAFICO']) if i + 1 < len(data) else ''
        
        # Extraemos el editorial de la fila que sigue al título
        editorial = str(data.loc[i + 2, 'ASIENTO BIBLIOGRAFICO']) if i + 2 < len(data) else ''
        
        # Extraemos el ISBN de la fila que sigue al editorial
        isbn = str(data.loc[i + 3, 'ASIENTO BIBLIOGRAFICO']) if i + 3 < len(data) else ''
        
        # Añadimos el registro a la lista
        extracted_records.append({"Nº": numero, "Autor": author, "Título": title, "Editorial": editorial, "ISBN": isbn})

# Crear DataFrame de los registros extraídos
extracted_df = pd.DataFrame(extracted_records)

# Guardar DataFrame en un archivo CSV
extracted_df.to_csv("ListaDeLibros.csv", index=False)

# Mostrar los primeros registros
print(extracted_df.head())
```
